```python
import json
import pandas as pd
from python_submodule.general import info, warning, error, debug
```

```python
train_data = json.load(open("data/test_with_timestamps.json"))
test_data = json.load(open("data/test_with_timestamps.json"))
metadata = json.load(open("data/thinned_poi_metadata.json"))
```

```python
session_id = 0
train_dict = {"SessionId": [], "ItemId": [], "Time": []}

for e in train_data:
    if len(e["history_events"]) == 1:
        session_id += 1
        train_dict["SessionId"].append(session_id)
        train_dict["ItemId"].append(e["history_events"][0]["item_id"])
        train_dict["Time"].append(e["history_events"][0]["timestamp"])
    train_dict["SessionId"].append(session_id)
    train_dict["ItemId"].append(e["target_event"]["item_id"])
    train_dict["Time"].append(e["target_event"]["timestamp"])

train_df = pd.DataFrame(train_dict)
```

```python
session_id = 0
test_dict = {"SessionId": [], "ItemId": [], "Time": []}

for e in test_data:
    if len(e["history_events"]) == 1:
        session_id += 1
        test_dict["SessionId"].append(session_id)
        test_dict["ItemId"].append(e["history_events"][0]["item_id"])
        test_dict["Time"].append(e["history_events"][0]["timestamp"])
    test_dict["SessionId"].append(session_id)
    test_dict["ItemId"].append(e["target_event"]["item_id"])
    test_dict["Time"].append(e["target_event"]["timestamp"])

test_df = pd.DataFrame(test_dict)
```

```python
train_df["Time"] = pd.to_datetime(train_df["Time"]).astype('int64') // 10**9
test_df["Time"] = pd.to_datetime(test_df["Time"]).astype('int64') // 10**9
```
